```python
#                       Question#1
students=[]
def calculate(py,ai,math):
    total=py+ai+math
    percentage=total/300*100
    if percentage>= 80:
        grade = "A"
    elif percentage>=70:
        grade="B"
    elif percentage>=60:
        grade="C"
    elif percentage>=50:
        grade="D"
    else:
        grade="F"
    return total , percentage,grade
   
#input for students
for i in range(5):
    print("Student:",i+1)
    
    name=input("Enter your name:")
    roll=input("Enter roll no:")
    py=int(input("Python marks:"))
    ai=int(input("AI marks:"))
    math=int(input("Math marks:"))

    total,percentage,grade=calculate(py,ai,math)
    student={"Name":name,
             "Rollno":roll,
             "Total":total,
             "Percentage":percentage,
             "Grade":grade
            }

    students.append(student)

    print("=====Student result======")

    for s in students:
        print("Name:",s["Name"])
        print("Rollno:",s["Rollno"])
        print("Total:",s["Total"])
        print("Percentage:",s["Percentage"])
        print("Grade:",s["Grade"])
        print("------")
#TOPPER
    topper =students[0]
    for s in students:
        if s["Percentage"]>topper["Percentage"]:
            topper=s
            print("=====TOPPER=====")
    print("Topper:",topper["Name"])
    print("Percentage:",topper["Percentage"])

#Save in file
with open("result.txt","w")as file:
    for s in students:
        file.write("Name:"+s["Name"])
        file.write("Roll no:"+s["Rollno"])
        file.write("Total:"+str(s["Total"]))
        file.write("Percentage"+str(s["Percentage"]))
        file.write("Grade:"+s["Grade"])
       






```

```python
#                             Question #2
Sentence=input("Enter a sentence")
words=Sentence.split()
total_words=len(words)
print("Total words:",total_words)
ch=len(Sentence)
print("Total characters:",ch)
longest_word=max(words,key=len)
print("Longest words:",longest_word)
shortest_word=min(words,key=len)
print("Shortest words:",shortest_word)
#VOWELS
vowels="aeiou"
v=sum(Sentence.count(ch)for ch in vowels)
print("Total numbers of vowels:",v)
total_words=Sentence.split()

#list
l=list()
for i in total_words:
    if i not in l:
        l.append(i)

g=str(l)
print(l)
sorted_words=sorted(words)
print("Sorted sentence",sorted_words)
#save in file
f=open("Analysis.txt","a")
f.write("Total words:"+str(total_words)+"\n")
f.write("Longest words:"+longest_word+"\n")
f.write("Total characters:"+str(ch)+"\n")
f.write("Total number of vowels:"+str(v)+"\n")
f.write("Set of unique words:"+g+"\n")
f.write("Sorted sentence:"+str(sorted_words)+"\n")
f.close()



   
```

```python
#                             Question#04
def addition(a,b):
    return a+b
def subtraction(a,b):
    return a-b
def multiply(a,b):
    return a*b
def divide(a,b):
    return a/b
def modulus(a,b):
    return a%b
while True:
    num1=float(input("Enter the first number:"))
    num2=float(input("Enter the second number:"))
    operation=input("Enter operation(+,-,*,/,%)")
    if operation=="+":
        result=addition(num1,num2)
    elif operation=="-":
        result=subtraction(num1,num2)
    elif operation=="*":
        result=multiply(num1,num2)
    elif operation=="/":
        result=divide(num1,num2)
    elif operation=="%":
        result=modulus(num1,num2)
    else:
        print("Invalid operation")
        continue
    print("Result:",result)

    again=input("Do you want another calculation(yes/no):")
    if again=="no":
        break
print("Calculator closed.")
    



    
```

```python
#                                   QUESTION#05
menu={
    "1":"Burger",
    "2":"Pizza",
    "3":"Fries",
    "4":"Drink"
}
prices={
    "Burger":500,
    "Pizza":800,
    "Fries":250,
    "Drink":150
}
def cal_price(item,quantity):
    return prices[item]*quantity

print("=======MENU=========")
print("1.Burger Rs.500")
print("2.Pizza  Rs.800")
print("3.Fries  Rs.250")
print("4.Drink  Rs.150")

bill=0
order=""
while True:
    choice = input("Select item number:")
    if choice in menu:
        item=menu[choice]
        quantity=int(input("Enter quantity:"))
        price=cal_price(item,quantity)
        bill=bill+price
        order=order+item+"x"+str(quantity)+"=RS."+str(price)+"\n"

    else:
        print("Wrong choice")

    more=input("Want to order more?(yes/no)")
    if more!="Yes":
     break
        

print("======YOUR BILL========")
print(order)
print("Total bill=Rs.",bill)

file=open("Order.txt","w")
file.write("======ORDER======\n")
file.write(order)
file.write("Total Bill=RS."+str(bill)+"/n")
file.close()

print("Order saved in order.txt")

```
